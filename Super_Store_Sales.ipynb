# Import Libraries

```python
import pandas as pd
import seaborn as sns
import sqlite3
import matplotlib.pyplot as plt
```

## 1. Load CSV and create database

```python
df = pd.read_csv(
    'C:/Users/Chandana S R/CANDY/Data Analytics Docs/SuperStoreData/Superstore_Sales.csv',
    encoding='latin1'
)

conn = sqlite3.connect('sales_data.db')
df.to_sql('Superstore_Sales', conn, if_exists='replace', index=False)
conn.close()
print('✅ Database file created and sales table added successfully.\n')
```

## 2. Connect and run SQL query

```python
conn = sqlite3.connect('sales_data.db')
query = '''
SELECT
    `Product ID`,
    SUM(Quantity) AS TotalQuantity,
    SUM(Sales) AS TotalRevenue
FROM Superstore_Sales
GROUP BY `Product ID`
'''
df = pd.read_sql_query(query, conn)
conn.close()
```

## 3. Print sales summary

```python
print("🧾 Sales Summary:\n")
for _, row in df.iterrows():
    print(f"{row['Product ID']}: {row['TotalQuantity']} units sold, ${row['TotalRevenue']:.2f} in revenue")
```

## 4. Plot top 10 products by revenue

```python
top_products = df.sort_values(by='TotalRevenue', ascending=False).head(10)
sns.set(style='whitegrid')

plt.figure(figsize=(12, 6))
bars = plt.bar(top_products['Product ID'], top_products['TotalRevenue'], color='teal')
```

## Add value labels on top

```python
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width() / 2, height + 100, f"${height:,.0f}",
             ha='center', va='bottom', fontsize=8)
```

## Final formatting

```python
plt.title("Top 10 Products by Revenue", fontsize=14)
plt.xlabel('Product ID', fontsize=12)
plt.ylabel('Revenue ($)', fontsize=12)
plt.xticks(rotation=45, ha='right', fontsize=9)

plt.tight_layout()
plt.savefig("top_10_products_by_revenue.png", dpi=300, bbox_inches='tight')
plt.show()
```
